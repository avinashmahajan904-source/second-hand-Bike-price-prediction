```python
log =  ['riding_range']

exp = ['kilometer']

ordinal = ['owner']

normal = ['brand_name','rear_brake_type','front_brake_type','insurance','tyre_type','city']

plain = ['reserve_fuel_capacity','fuel_tank_capacity','cylinders','mileage_owner_reported','bike_age']
```

```python
feature = log +  normal + plain + exp + ordinal
```

```python
x=df[feature]
```

```python
x.head()
```

```python
y = df['price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test ,y_train ,y_test = train_test_split(x , y ,test_size = 0.2 ,random_state = 42)
```

```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.preprocessing import  FunctionTransformer,StandardScaler,OneHotEncoder,OrdinalEncoder
```

```python
# log function pipeline 
log_pipeline = Pipeline(steps = [
    ('log',FunctionTransformer(np.log)),
    ('standarscaler',StandardScaler())
])

# exp function pipeline
exp_pipeline = Pipeline(steps = [
    ('exp',FunctionTransformer(np.exp)),
    ('standardscaler',StandardScaler())
])

#standard Scaler pipeline
Plain_pipeline = Pipeline(steps = [
    ('standarscaler',StandardScaler())
])

# One Hot Encoding Pipeline 
one_hot_pipeline = Pipeline(steps = [
    ('one_hot_encoder',OneHotEncoder(handle_unknown = 'ignore'))
])

# ordinal pipeline 
ordinal_pipeline = Pipeline(steps = [
    ('ordinal_encoder', OrdinalEncoder(categories=[['First','Second','Third','Fourth','4 or More']]))
])
```

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor  = ColumnTransformer(transformers = [
    ('log',log_pipeline,log),
    ('exp',exp_pipeline,exp),
    ('plain',Plain_pipeline,plain),
    ('one_hot_encoder',one_hot_pipeline,normal),
    ('ordinal',ordinal_pipeline,ordinal)
])
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr_pipeline = Pipeline(steps = [
    ('preprocessor',preprocessor),
    ('linear model',LinearRegression())
])
```

```python
lr_pipeline.fit(x_train ,y_train)
```

```python
y_pred = lr_pipeline.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
clf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(n_estimators=100,random_state=42))
])
```

```python
clf_pipeline.fit(x_train,y_train)
```

```python
y_pred_clf = clf_pipeline.predict(x_test)
```

```python
r2_score(y_test,y_pred_clf)
```

```python
y_pred_train_clf = clf_pipeline.predict(x_train)
```

```python
r2_score(y_train,y_pred_train_clf)
```

```python
from xgboost import XGBRegressor
```

```python
xgb=Pipeline(steps = [
    ('preprocessor',preprocessor),
    ('xgb',XGBRegressor(n_estimators=100,learning_rate=0.07,random_state=42))
])
```

```python
xgb.fit(x_train,y_train)
```

```python
y_pred_xgb=xgb.predict(x_test)
```

```python
r2_score(y_test,y_pred_xgb)
```

```python
y_train_xgb=xgb.predict(x_train)
```

```python
r2_score(y_train,y_train_xgb)
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'xgb__n_estimators': [50,100, 200, 300],
    'xgb__max_depth': [3, 5, 7,9],
    'xgb__learning_rate': [0.01, 0.1, 0.2,0.07]
}

grid_search = GridSearchCV(estimator= xgb, param_grid = param_grid, cv=5, n_jobs=-1)
grid_search.fit(x_train, y_train)

print(grid_search.best_params_)
best_xgb = grid_search.best_estimator_
```

```python
y_pred_xgb_best = best_xgb.predict(x_test)
```

```python
r2_score(y_test, y_pred_xgb_best)
```

```python
y_train_xgb_best = best_xgb.predict(x_train)
```

```python
r2_score(y_train , y_train_xgb_best)
```
